```python
# CHESSONOMICS: Material Calculator

piece_values = {
    "Pawn": 1,
    "Knight": 3,
    "Bishop": 3,
    "Rook": 5,
    "Queen": 9
}

print("CHESS MATERIAL CALCULATOR")
print("-------------------------")

white_material = (
    8 * piece_values["Pawn"]
    + 2 * piece_values["Knight"]
    + 2 * piece_values["Bishop"]
    + 2 * piece_values["Rook"]
    + piece_values["Queen"]
)

black_material = (
    8 * piece_values["Pawn"]
    + 2 * piece_values["Knight"]
    + 2 * piece_values["Bishop"]
    + 2 * piece_values["Rook"]
    + piece_values["Queen"]
)

print("White material:", white_material)
print("Black material:", black_material)

print("Material difference:",
      white_material - black_material)
```

```python
# CHESSONOMICS: Custom Material Calculator

piece_values = {
    "Pawn": 1,
    "Knight": 3,
    "Bishop": 3,
    "Rook": 5,
    "Queen": 9
}

def calculate_material(position):
    total = 0

    for piece, count in position.items():
        total += piece_values[piece] * count

    return total


# Example hypothetical position

white_pieces = {
    "Pawn": 6,
    "Knight": 2,
    "Bishop": 1,
    "Rook": 2,
    "Queen": 1
}

black_pieces = {
    "Pawn": 5,
    "Knight": 2,
    "Bishop": 2,
    "Rook": 1,
    "Queen": 1
}

white_total = calculate_material(white_pieces)
black_total = calculate_material(black_pieces)

print("CHESSONOMICS MATERIAL ANALYSIS")
print("--------------------------------")

print("White material:", white_total)
print("Black material:", black_total)

difference = white_total - black_total

print("Material difference:", difference)

if difference > 0:
    print("White has a material advantage.")
elif difference < 0:
    print("Black has a material advantage.")
else:
    print("Material is equal.")
```

```python
# CHESSONOMICS: Opportunity Cost Calculator

piece_values = {
    "Pawn": 1,
    "Knight": 3,
    "Bishop": 3,
    "Rook": 5,
    "Queen": 9
}

# Hypothetical chess exchange

piece_given = "Rook"
piece_received = "Bishop"

cost = piece_values[piece_given]
benefit = piece_values[piece_received]

net_value = benefit - cost

print("OPPORTUNITY COST ANALYSIS")
print("-------------------------")

print("Piece given:", piece_given)
print("Piece received:", piece_received)

print("Value sacrificed:", cost)
print("Value gained:", benefit)

print("Net material change:", net_value)

if net_value > 0:
    print("Material gain.")
elif net_value < 0:
    print("Material loss.")
else:
    print("Material exchange is equal.")
```

```python
# CHESSONOMICS: Expected Value Calculator

def expected_value(probability, success, failure):
    return (
        probability * success
        + (1 - probability) * failure
    )

# Safe strategy
safe_ev = expected_value(0.80, 3, -1)

# Risky strategy
risky_ev = expected_value(0.40, 10, -3)

print("RISK AND EXPECTED VALUE ANALYSIS")
print("--------------------------------")

print("Safe strategy EV:", round(safe_ev, 2))
print("Risky strategy EV:", round(risky_ev, 2))

if safe_ev > risky_ev:
    print("Safe strategy has higher expected value.")
elif risky_ev > safe_ev:
    print("Risky strategy has higher expected value.")
else:
    print("Both strategies have equal expected value.")
```

```python
# CHESSONOMICS: Risk Measurement

def calculate_risk(probability, success, failure):

    ev = (
        probability * success
        + (1 - probability) * failure
    )

    variance = (
        probability * (success - ev) ** 2
        + (1 - probability) * (failure - ev) ** 2
    )

    standard_deviation = variance ** 0.5

    return ev, variance, standard_deviation


# Safe strategy
safe = calculate_risk(0.80, 3, -1)

# Risky strategy
risky = calculate_risk(0.40, 10, -3)

print("RISK COMPARISON")
print("----------------")

print("SAFE STRATEGY")
print("Expected Value:", round(safe[0], 2))
print("Variance:", round(safe[1], 2))
print("Standard Deviation:", round(safe[2], 2))

print("\nRISKY STRATEGY")
print("Expected Value:", round(risky[0], 2))
print("Variance:", round(risky[1], 2))
print("Standard Deviation:", round(risky[2], 2))

if safe[2] < risky[2]:
    print("\nSafe strategy has lower variability.")
else:
    print("\nRisky strategy has lower variability.")
```

```python
# CHESSONOMICS: Game Theory

payoff_matrix = {
    "Defensive": {
        "Defensive": (3, 3),
        "Aggressive": (1, 5)
    },

    "Aggressive": {
        "Defensive": (5, 1),
        "Aggressive": (2, 2)
    }
}

print("GAME THEORY PAYOFF MATRIX")
print("-------------------------")

for strategy_a, outcomes in payoff_matrix.items():

    for strategy_b, payoff in outcomes.items():

        print(
            strategy_a,
            "vs",
            strategy_b,
            "=>",
            payoff
        )
```

```python
# CHESSONOMICS: Nash Equilibrium Finder

strategies = ["Defensive", "Aggressive"]

payoff_matrix = {
    "Defensive": {
        "Defensive": (3, 3),
        "Aggressive": (1, 5)
    },
    "Aggressive": {
        "Defensive": (5, 1),
        "Aggressive": (2, 2)
    }
}

equilibria = []

for a in strategies:
    for b in strategies:

        current_a, current_b = payoff_matrix[a][b]

        # Check whether Player A can improve
        best_a = max(
            payoff_matrix[alternative_a][b][0]
            for alternative_a in strategies
        )

        # Check whether Player B can improve
        best_b = max(
            payoff_matrix[a][alternative_b][1]
            for alternative_b in strategies
        )

        if current_a == best_a and current_b == best_b:
            equilibria.append((a, b))

print("NASH EQUILIBRIUM ANALYSIS")
print("-------------------------")

if equilibria:
    for equilibrium in equilibria:
        print("Nash Equilibrium:", equilibrium)
else:
    print("No pure-strategy Nash Equilibrium found.")
```

```python
# CHESSONOMICS: Result Percentage Analysis

import pandas as pd

games = {
    "Result": [
        "Win", "Draw", "Loss", "Win", "Draw",
        "Win", "Loss", "Win", "Draw", "Win"
    ]
}

df = pd.DataFrame(games)

total_games = len(df)

result_counts = df["Result"].value_counts()

print("CHESS RESULT PERCENTAGE ANALYSIS")
print("--------------------------------")

for result in ["Win", "Draw", "Loss"]:

    count = result_counts.get(result, 0)

    percentage = (count / total_games) * 100

    print(
        f"{result}: {count} games "
        f"({percentage:.1f}%)"
    )

print("--------------------------------")
print("Total games:", total_games)
```
